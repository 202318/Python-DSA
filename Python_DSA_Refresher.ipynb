# Question 1: Converting Days to Seconds

Write a program that asks the user for a number of days and displays the equivalent number of seconds.

```python
# Ask the user to enter the number of days
days = int(input("Enter the number of days: "))

# Convert the number of days into seconds
seconds = days * 24 * 60 * 60

# Display the result
print("Number of seconds:", seconds)
```

# Question 2: Volume of a Sphere

Write a program that asks the user to enter the radius and calculates the volume of a sphere.

```python
# Import the math module so that we can use pi
import math

# Ask the user to enter the radius
radius = float(input("Enter the radius: "))

# Calculate the volume of the sphere
volume = (4 / 3) * math.pi * (radius ** 3)

# Display the calculated volume
print("Volume of the sphere:", volume)
```

# Question 3: Area and Perimeter of a Square

Using functions, write a program that asks the user for the side length of a square and calculates its area and perimeter.

```python
# Function to calculate the area of a square
def calculate_area(side):
    return side ** 2

# Function to calculate the perimeter of a square
def calculate_perimeter(side):
    return 4 * side

# Ask the user to enter the side length
side = float(input("Enter the side length of the square: "))

# Calculate the area and perimeter using the functions
area = calculate_area(side)
perimeter = calculate_perimeter(side)

# Display the results
print("Area of the square:", area)
print("Perimeter of the square:", perimeter)
```

# Question 4: Uppercase or Lowercase Character

Write a program using a function that determines whether a character entered by the user is uppercase or lowercase.

```python
# Function to determine whether a character is uppercase or lowercase
def check_case(character):
    if character.isupper():
        return "Uppercase"
    elif character.islower():
        return "Lowercase"
    else:
        return "Not an alphabetic character"

# Ask the user to enter a character
character = input("Enter a character: ")

# Call the function and display the result
result = check_case(character)
print("The character is:", result)
```

# Question 5: Converting Pseudocode to Python

Write the Python program equivalent of the given pseudocode.

```python
# Set the initial values of x and y
x = 0
y = 20

# Repeat the calculations
while True:
    y = y - 4
    x = x + (2 / y)

    # Stop the loop when y is less than 6
    if y < 6:
        break

# Display the final value of x
print("x =", x)
```

# Question 6: Arrays and Average

Write a Python program that uses a loop to input 5 values into an array and calculates and displays their average.

```python
# Create an empty list to store the five values
values = []

# Use a loop to ask the user to enter 5 values
for i in range(5):
    value = float(input("Enter a value: "))
    values.append(value)

# Calculate the average of the five values
average = sum(values) / len(values)

# Display the values and their average
print("Values entered:", values)
print("Average:", average)
```
